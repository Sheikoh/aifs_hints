```python
# define module, exo
def fetch_help(module, exercise, task, content="hint"):
    if content == "hint":
        url = 'https://raw.githubusercontent.com/Sheikoh/aifs_hints/refs/heads/main/src/hints_aifs.csv'

    else:
        content="todo"
        url = 'https://raw.githubusercontent.com/Sheikoh/aifs_todo/refs/heads/main/src/hints_aifs.csv'
    hints = pd.read_csv(url, index_col=0)
    hints_exo = hints.loc[(hints["module"] == module) & (hints["exercise"] == exercise) & (hints["task"] == task)]
    return print(hints_exo.iloc[0]["todo"])
```
